# Model C.1 — More Training with Early Stopping 

This experiment trains Model C for a maximum of 15 to 30 training cycles, adding an early stop feature with a five-cycle waiting period. The model's structure and organization settings remain unchanged. The aim of this test is to determine whether Model C requires additional training time to stabilize and give better resulte.

```python
from pathlib import Path
import sys
import json
import time
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
)

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
import json
import time
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from sklearn.metrics import confusion_matrix, classification_report
from project_utils import (
    SEED,
    EPOCHS,
    TARGET_CLASSES,
    RESULTS_DIR,
    FIGURES_DIR,
    ARTIFACTS_DIR,
    set_seed,
    load_datasets,
    build_cnn,
)
```

```python
set_seed()

train_ds, val_ds, test_ds = load_datasets()

for images, labels in train_ds.take(1):
    print("Images shape:", images.shape)
    print("Labels shape:", labels.shape)
```

```python
# Model C configuration: regularized CNN with Data Augmentation, Batch Normalization, and Dropout
model_c_config = {
    "name": "model_c_early_stopping",
    "filters": [32, 64, 128],
    "kernel_size": 3,
    "augmentation": True,
    "batch_norm": True,
    "dense_units": 64,
    "dropout": 0.3,
}
```

```python
tf.keras.backend.clear_session()
set_seed()

model_c = build_cnn(
    num_classes=len(TARGET_CLASSES),
    config=model_c_config,
)

model_c.summary()

model_c_params = model_c.count_params()

print(f"Total parameters: {model_c_params:,}")
```

```python
#model compile
model_c.compile(optimizer=tf.keras.optimizers.Adam(), loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],)
```

```python
callbacks_c = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True,),]
```

```python
start_time_c = time.time()

history_c = model_c.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    callbacks=callbacks_c,
)

train_time_sec_c = (time.time() - start_time_c)
epochs_completed_c = len(history_c.history["loss"])
print(
    f"\nTraining time: {train_time_sec_c:.1f} "
    f"sec for {epochs_completed_c} epochs")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].plot(history_c.history["loss"], label="train")
axes[0].plot(history_c.history["val_loss"], label="val")
axes[0].set_title("Model C Early Stopping Loss")
axes[0].set_xlabel("Epoch")
axes[0].legend()

axes[1].plot(history_c.history["accuracy"], label="train")
axes[1].plot(history_c.history["val_accuracy"], label="val")
axes[1].set_title("Model C Early Stopping Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].legend()

plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, "model_c_early_stopping_curves.png"), dpi=120)
plt.show()
```

```python
test_loss_c , test_acc_c = model_c.evaluate(test_ds, verbose=0)
print(f"Model C test accuracy: {test_acc_c:.4f} | test loss: {test_loss_c:.4f}")

y_true_c, y_pred_c = [], []
for imgs, labels in test_ds:
    probs = model_c.predict(imgs, verbose=0)
    y_true_c.extend(labels.numpy().tolist())
    y_pred_c.extend(np.argmax(probs, axis=1).tolist())

cm = confusion_matrix(y_true_c, y_pred_c)
fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(cm, cmap="Blues")
ax.set_xticks(range(len(TARGET_CLASSES))); ax.set_xticklabels(TARGET_CLASSES, rotation=90)
ax.set_yticks(range(len(TARGET_CLASSES))); ax.set_yticklabels(TARGET_CLASSES)
ax.set_xlabel("Predicted"); ax.set_ylabel("True")
ax.set_title("Model C Early Stopping - Confusion Matrix (test set)")
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(j, i, cm[i, j], ha="center", va="center",
                color="white" if cm[i, j] > cm.max() / 2 else "black", fontsize=7)
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_DIR, "model_c_early_stopping_confusion_matrix.png"), dpi=120)
plt.show()

print(classification_report(y_true_c, y_pred_c, target_names=TARGET_CLASSES))
```

```python
model_c.save(os.path.join(ARTIFACTS_DIR,"model_c_early_stopping.keras",))

history_out_c = {
    "model": "model_c_early_stopping",
    "config": model_c_config,
    "epochs": int(epochs_completed_c),
    "train_time_sec": float(train_time_sec_c),
    "n_params": int(model_c_params),
    "test_accuracy": float(test_acc_c),
    "test_loss": float(test_loss_c),
    "history": {
        k: [float(v) for v in vals]
        for k, vals in history_c.history.items()
    },}

with open(os.path.join(RESULTS_DIR,"model_c_early_stopping_history.json",),"w",) as f:
    json.dump(history_out_c,f,indent=2,)

print("Saved model to",os.path.join(ARTIFACTS_DIR,"model_c_early_stopping.keras",),)

print("Saved training history to",os.path.join(RESULTS_DIR,"model_c_early_stopping_history.json",),)
```

## Findings

Overall, Model C with extended training achieved a test accuracy of 82.39%, a test loss of 0.5374, and a macro F1-score of 0.83. This was a substantial improvement over the original Model C, indicating that the regularized architecture required more training time to stabilize.
Yet, the experiment was excluded from the main comparison because it exceeded the shared 15-epoch constraint.
